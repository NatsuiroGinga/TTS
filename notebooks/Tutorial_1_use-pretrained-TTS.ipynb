## 性能测试

````python
meta_path = '/content/seedtts_testset/zh/meta.lst'  # 根据实际路径调整
with open(meta_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

# 解析meta文件每一行，按'|'分隔
samples = []
for line in lines:
    fields = line.strip().split('|')
    if len(fields) < 4:
        continue
    sample = {
        # 待合成文本的音频文件名
        'synthesis_audio': fields[0],
        # 提示文本
        'prompt_text': fields[1],
        # 提示音频路径
        'prompt_audio': fields[2],
        # 待合成文本
        'synthesis_text': fields[3]
        # 第五字段ground_truth如果存在可以忽略，因为仅进行TTS
    }
    samples.append(sample)

print(f"共解析到 {len(samples)} 行数据")

# 打印前5行
print("\n打印前5行")
for sample in samples[:5]:
    print(sample)

# 输出第一行
print("\n输出第一行语音和文本")
print("prompt_text:", samples[0]['prompt_text'])
print("播放prompt_audio")
Audio("/content/seedtts_testset/zh/" + samples[0]['prompt_audio'], rate=24_000, autoplay=True)
````

````python
nvmlInit()
# 获取第一个 GPU 设备的句柄（若有多GPU，可遍历或指定索引）
handle = nvmlDeviceGetHandleByIndex(0)
# Get device
device = "cuda" if torch.cuda.is_available() else "cpu"
# Init TTS
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
````

````python
inference_times = []
rtf_values = []
cpu_usages = []
mem_usages = []
pesq_scores = []
ground_truths = []
gpu_utils = []   # 用于记录 GPU 利用率
mem_utils = []   # 用于记录显存利用率（百分比）

for sample in samples[:100]:
    # 取出文本
    text = sample['synthesis_text']

    # 资源
    process = psutil.Process(os.getpid())
    cpu_before = psutil.cpu_percent(interval=None)
    mem_before = process.memory_info().rss / (1024 ** 2)  # 转换为 MB

    utilization_before = nvmlDeviceGetUtilizationRates(handle)
    memory_info_before = nvmlDeviceGetMemoryInfo(handle)
    gpu_util_before = utilization_before.gpu
    mem_used_before = memory_info_before.used / (1024**2)  # 转换为MB

    # 时间
    start_time = time.time()
    # 生成语音
    wav = tts.tts(text=text, speaker="Daisy Studious", language="zh-cn")

    # 时间
    end_time = time.time()
    elapsed = end_time - start_time
    inference_times.append(elapsed)
    audio_length = len(wav) / 24000
    rtf = elapsed / audio_length if audio_length > 0 else None
    print(f"audio_length:{audio_length:.2f}, elapsed:{elapsed:.2f}, rtf:{rtf}")
    rtf_values.append(rtf)

    # 资源
    cpu_after = psutil.cpu_percent(interval=None)
    mem_after = process.memory_info().rss / (1024 ** 2)
    cpu_usages.append((cpu_before + cpu_after) / 2)
    mem_usages.append((mem_before + mem_after) / 2)

    utilization_after = nvmlDeviceGetUtilizationRates(handle)
    memory_info_after = nvmlDeviceGetMemoryInfo(handle)
    gpu_util_after = utilization_after.gpu
    mem_used_after = memory_info_after.used / (1024**2)  # 转换为MB

    # 计算平均利用率和平均显存占用
    avg_gpu_util = (gpu_util_before + gpu_util_after) / 2
    avg_mem_used = (mem_used_before + mem_used_after) / 2
    gpu_utils.append(avg_gpu_util)
    mem_utils.append(avg_mem_used)
````

````python
nvmlShutdown() # 关闭 NVML
````

### 推理速度测试

````python
# 计算平均 RTF 值
average_rtf = np.mean([rtf for rtf in rtf_values if rtf is not None])
print("Average RTF:", average_rtf)

# RTF 折线图
plt.figure(figsize=(10,5))
plt.plot(range(len(rtf_values)), rtf_values, marker='o')
plt.title("RTF")
plt.xlabel("Sample ID")
plt.ylabel("RTF")
plt.grid(True)
# 绘制平均值的水平线
plt.axhline(y=average_rtf, color='r', linestyle='--', label=f'Average RTF: {average_rtf:.3f}')
plt.legend()
plt.show()
````

### 资源占用测试

#### CPU

````python
# 计算平均值
average_cpu = np.mean(cpu_usages)
average_mem = np.mean(mem_usages)
# 输出

# CPU 使用率折线图
plt.figure(figsize=(10,4))
plt.plot(range(len(cpu_usages)), cpu_usages, marker='o', label='CPU Usage (%)')
plt.title("CPU Usage During Inference")
plt.xlabel("Sample ID")
plt.ylabel("CPU Usage (%)")
# 画出平均值的线
plt.axhline(y=average_cpu, color='r', linestyle='--', label=f'Average CPU: {average_cpu:.3f}%')
plt.grid(True)
plt.legend()
plt.show()
````

#### GPU

````python
# 计算平均值
average_gpu_util = np.mean(gpu_utils)
average_mem_util = np.mean(mem_utils)

plt.figure(figsize=(10,4))
plt.plot(range(len(gpu_utils)), gpu_utils, marker='o', label='GPU Usage (%)')
plt.title("GPU Usage (%)")
plt.xlabel("Sample ID")
plt.ylabel("GPU Usage (%)")
# 画平均线
plt.axhline(y=average_gpu_util, color='r', linestyle='--', label=f'Average GPU: {average_gpu_util:.3f}%')
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(10,4))
plt.plot(range(len(mem_utils)), mem_utils, marker='o', color='orange', label='Memory Usage (MB)')
plt.title("GPU Memory Usage (MB)")
plt.xlabel("Sample ID")
plt.ylabel("Memory Usage (MB)")
# 画平均线
plt.axhline(y=average_mem_util, color='r', linestyle='--', label=f'Average Memory: {average_mem_util:.3f} MB')
plt.grid(True)
plt.legend()
plt.show()
````
